```python
import matplotlib.pyplot as plt # Importa la librería Matplotlib para la creación de gráficos
```

```python
plt.figure(figsize=(12,8)) # Crea una nueva figura con un tamaño específico
plt.hist(data['CREDIT'], bins = 500) # Genera un histograma de la columna 'CREDIT' con 500 barras
plt.show() # Muestra el gráfico
```

```python
plt.figure(figsize=(12,8)) # Crea una nueva figura con un tamaño específico
plt.hist(data['CREDIT'], bins = 500, color='#5dc1b9') # Genera un histograma con color personalizado
plt.title("Histograma de créditos") # Añade un título al histograma
plt.show() # Muestra el gráfico
```

```python
plt.figure(figsize=(12,8)) # Crea una nueva figura con un tamaño específico
plt.hist(data['CREDIT'], bins=500, density = True, cumulative= True) # Genera un histograma acumulativo y normalizado
plt.show() # Muestra el gráfico
```

```python
plt.figure(figsize=(12,8)) # Crea una nueva figura con un tamaño específico
plt.hist(data['CREDIT'], bins = 500, range=(100000,2000000)) # Genera un histograma con un rango específico en el eje X
plt.show() # Muestra el gráfico
```

```python
plt.figure(figsize=(12,8)) # Crea una nueva figura con un tamaño específico
plt.hist(data['CREDIT'], bins = 200, histtype= 'step', color = 'purple') # Genera un histograma de tipo 'step' (solo contornos)
plt.show() # Muestra el gráfico
```

```python
plt.figure(figsize=(12,8)) # Crea una nueva figura con un tamaño específico
plt.hist(data['CREDIT'], bins = 200, histtype= 'stepfilled', color= 'pink') # Genera un histograma de tipo 'stepfilled' (relleno)
plt.show() # Muestra el gráfico
```

```python
plt.figure(figsize=(12,8)) # Crea una nueva figura con un tamaño específico
plt.hist(data['CREDIT'], bins = 500, orientation='horizontal', color='brown') # Genera un histograma horizontal
plt.show() # Muestra el gráfico
```

```python
plt.figure(figsize=(12,8)) # Crea una nueva figura con un tamaño específico
plt.hist(data['CREDIT'], bins = 500, color= '#aa8caf',log = True) # Genera un histograma con escala logarítmica en el eje Y
plt.show() # Muestra el gráfico
```

```python
from matplotlib.lines import lineStyles # Importa lineStyles (aunque no se usa directamente en este snippet)
# PARTE 1: LA ANATOMIA MATEMATICA Y VISUAL DEL BOXPLOT (Usando la Edad)

# 1. Calculos matematicos paso a paso
Q1 = data["FAM_MEMBERS"].quantile(0.25) # Calcula el primer cuartil (Q1) de la columna 'FAM_MEMBERS'
Mediana = data["FAM_MEMBERS"].median() # Calcula la mediana de la columna 'FAM_MEMBERS'
Q3 = data["FAM_MEMBERS"].quantile(0.75) # Calcula el tercer cuartil (Q3) de la columna 'FAM_MEMBERS'
IQR = Q3 - Q1 # Calcula el Rango Intercuartílico (IQR)
limite_sup = Q3 + 1.5 * IQR # Calcula el límite superior para identificar outliers
limite_inf = Q1 - 1.5 * IQR # Calcula el límite inferior para identificar outliers

# Imprimimos el "guion" para la clase
print("--- RADIOGRAFIA DE NUMERO DE MIEMBROS FAMILIARES EN LA EMPRESA ---")
print(f"1. El Q1 (Inicio de la caja) es: {Q1:.0f} miembros familiares. El 25% de la empresa es menor a esto.")
print(f"2. El Q3 (Fin de la caja) es: {Q3:.0f} miembros familiares. El 75% de la empresa es menor a esto.")
print(f"3. La masa principal (el IQR o tamaño de la caja) esta entre {Q1:.0f} y {Q3:.0f} miembros familiares.")
print(f"4. ¡Alerta de Outlier! Segun las reglas matematicas, cualquier edad mayor a {limite_sup:.0f} miembros familiares es atipica.\n")

#2. Visualizacion Didactica: Dibujando las formulas
plt.figure(figsize=(14, 4)) # Crea una nueva figura para el gráfico con un tamaño específico
sns.boxplot(x=data["FAM_MEMBERS"], color="lightgreen", width=0.4) # Genera un boxplot de la columna 'FAM_MEMBERS'

#3. Trazamos lineas verticales para conectar los numeros con el dibujo
plt.axvline(Q1, color="blue", linestyle="--", linewidth=2, label=f'Q1 ({Q1:.0f} miembros familiares)') # Dibuja una línea vertical para Q1
plt.axvline(Q3, color="purple", linestyle="--", linewidth=2, label=f'Q3 ({Q3:.0f} miembros familiares)') # Dibuja una línea vertical para Q3
plt.axvline(limite_sup, color="red", linestyle="--", linewidth=2, label=f'Frontera Outliers Sup ({limite_sup:.0f} miembros familiares)') # Dibuja una línea vertical para el límite superior de outliers

plt.title("Anatomia de un Boxplot: ¿De donde salen las lineas de la caja?", fontsize=14, fontweight="bold") # Añade un título al boxplot
plt.xlabel("Miembros familiares de los Empleados") # Etiqueta el eje X
plt.legend() # Muestra la leyenda del gráfico
plt.show() # Muestra el gráfico

print("EXPLICACIÓN")
print("")

print(f"5. La Mediana (el centro de la caja) es: {Mediana:.0f} miembros familiares. El 50% de la empresa es menor a esto.")
print("- Vean como la caja verde empieza exactamente en la linea azul (Q1) ytermina en la morada (Q3).")
print("- El bigote derecho intenta llegar hasta el dato mas lejano, pero choca contra la 'Pared Roja' (Limite Superior).")
```
